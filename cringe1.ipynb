```python
import nbformat
from nbconvert import PythonExporter

# Specify the input and output file paths
input_notebook = 'AgentSnake.ipynb'
output_script = 'AgentSnake.py'

# Read the notebook and parse it using nbformat
with open(input_notebook, 'r', encoding='utf-8') as f:
    notebook_content = f.read()
    notebook = nbformat.reads(notebook_content, as_version=4)

# Create a PythonExporter instance
exporter = PythonExporter()

# Export the notebook as a Python script
python_script, _ = exporter.from_notebook_node(notebook)

# Write the Python script to the output file
with open(output_script, 'w', encoding='utf-8') as f:
    f.write(python_script)
```
